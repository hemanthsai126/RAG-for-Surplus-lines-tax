# Massachusetts -- General Laws ch. 175 et seq. (Insurance) -> `data/massachusetts/ins_codes/*.md`

Massachusetts **insurance** statutes are in **General Laws Part I, Title XXII**, especially **Chapter 175** and related chapters through **178A** (official: [malegislature.gov](https://malegislature.gov/Laws/GeneralLaws)).

On **Justia**, use the **year-scoped** tree: **`/codes/massachusetts/{CODE_YEAR}/part-i/title-xxii/...`** (not `part-v`; insurance is under **Part I**). Section URLs look like **`.../chapter-175/section-1/`** or **`.../section-3a/`**.

**Cloudflare** often blocks plain **`httpx`**; this notebook uses **`curl_cffi`** with **`impersonate="chrome120"`** (same pattern as **`ins_ipynb/maryland.ipynb`**).

**Discovery:** BFS from the **Title XXII** index, following only **chapter** index pages whose slug is in the **insurance cluster (chapters 175--178 and letter suffixes such as 175A, 176J, 178A)**; every matching **`.../chapter-.../section-...`** link is collected (~**1,450** sections for 2024).

**Download:** text from **`div.primary-content`**, with Justia boilerplate stripped. Files are **`MGL_sec_<label>.md`** where **`<label>`** is **`{chapterSlug}-{sectionSlug}`** with hyphens mapped to underscores (e.g. `MGL_sec_175a_3a.md`).

**Config:** **`CODE_YEAR`** in paths. **`MAX_SECTIONS`** caps downloads (**0** = all). **`MAX_DISCOVERY_PAGES`** caps index fetches (**0** = no cap). **`REUSE_DISCOVERED_URLS`** skips discovery when **`_massachusetts_insurance_section_urls.txt`** exists.

Run with the **project root** as cwd (same as **`python -m app.ingest`**).

Then run **`python -m app.ingest`** from the project root.

```python
from __future__ import annotations

import re
import time
from collections import deque
from pathlib import Path
from urllib.parse import urljoin, urlparse

from bs4 import BeautifulSoup
from curl_cffi import requests as curl_requests

BASE = "https://law.justia.com"
CODE_YEAR = "2024"
PATH_PREFIX = f"/codes/massachusetts/{CODE_YEAR}/part-i/title-xxii"
TITLE_INDEX = f"{BASE}{PATH_PREFIX}/"

OUT_DIR = Path("data") / "massachusetts" / "ins_codes"
OUT_DIR.mkdir(parents=True, exist_ok=True)

CURL_IMPERSONATE = "chrome120"
REQUEST_DELAY_SEC = 0.12
TIMEOUT = 60.0

MAX_SECTIONS = 0
MAX_DISCOVERY_PAGES = 0

SKIP_EXISTING = True

DISCOVERED_LIST = OUT_DIR / "_massachusetts_insurance_section_urls.txt"
REUSE_DISCOVERED_URLS = True

section_path_re = re.compile(
    rf"{re.escape(PATH_PREFIX)}/chapter-([^/]+)/section-([^/]+)/?$",
    re.I,
)
```

```python
def curl_get(url: str) -> str:
    time.sleep(REQUEST_DELAY_SEC)
    r = curl_requests.get(url, impersonate=CURL_IMPERSONATE, timeout=TIMEOUT)
    r.raise_for_status()
    return r.text


def path_key(u: str) -> str:
    return urlparse(u).path.rstrip("/")


def insurance_chapter_slug(ch: str) -> bool:
    """Ch. 175 through 178A (Mass. GL insurance cluster: 175*, 176*, 177, 178, 178A)."""
    ch = ch.lower().split("/")[0]
    m = re.match(r"^(\d+)([a-z0-9-]*)$", ch)
    if not m:
        return False
    base = int(m.group(1))
    suf = m.group(2)
    if base < 175 or base > 178:
        return False
    if base in (175, 176):
        return True
    if base == 177:
        return suf == ""
    if base == 178:
        return suf in ("", "a")
    return False


def is_section_path(p: str) -> bool:
    return bool(section_path_re.search(p))


def discover_section_urls() -> list[str]:
    """BFS Title XXII; collect section URLs only for insurance chapters 175--178A."""
    start = TITLE_INDEX
    seen: set[str] = set()
    in_q: set[str] = {path_key(start)}
    q: deque[str] = deque([start])
    sections: set[str] = set()
    fetches = 0

    while q:
        if MAX_DISCOVERY_PAGES and fetches >= MAX_DISCOVERY_PAGES:
            break
        url = q.popleft()
        pk = path_key(url)
        in_q.discard(pk)
        if pk in seen:
            continue
        if is_section_path(pk):
            continue
        if not pk.startswith(PATH_PREFIX):
            continue
        seen.add(pk)
        html = curl_get(url)
        fetches += 1
        soup = BeautifulSoup(html, "html.parser")
        for a in soup.find_all("a", href=True):
            absu = urljoin(url, a["href"])
            p = path_key(absu)
            if f"/codes/massachusetts/{CODE_YEAR}/" not in p:
                continue
            if not p.startswith(PATH_PREFIX):
                continue
            if is_section_path(p):
                m = section_path_re.search(p)
                if m and insurance_chapter_slug(m.group(1)):
                    sections.add(BASE + p + "/")
            else:
                mch = re.search(r"/chapter-([^/]+)/?$", p, re.I)
                if mch and not insurance_chapter_slug(mch.group(1)):
                    continue
                if p in seen or p in in_q:
                    continue
                in_q.add(p)
                q.append(BASE + p + "/")

    return sorted(sections, key=lambda u: label_sort_key(statute_label_from_url(u)))


def statute_label_from_url(url: str) -> str:
    pk = path_key(url)
    m = section_path_re.search(pk)
    if not m:
        raise ValueError(f"cannot parse section from {url!r}")
    ch, sec = m.group(1), m.group(2)
    return f"{ch}-{sec}"


def label_sort_key(label: str) -> tuple:
    out: list[tuple[int, int | str]] = []
    for part in label.split("-"):
        if part.isdigit():
            out.append((0, int(part)))
        else:
            out.append((1, part.lower()))
    return tuple(out)


def label_to_filename(label: str) -> str:
    safe = label.replace("-", "_")
    return f"MGL_sec_{safe}.md"


def extract_primary_text(html: str) -> tuple[str, str]:
    soup = BeautifulSoup(html, "html.parser")
    title_el = soup.find("title")
    title_txt = title_el.get_text(strip=True) if title_el else ""
    pc = soup.select_one("div.primary-content")
    if pc:
        text = pc.get_text("\n", strip=True)
    else:
        main = soup.find("main") or soup.find("article")
        text = main.get_text("\n", strip=True) if main else soup.get_text("\n", strip=True)
    return title_txt, text


def strip_justia_boilerplate(text: str) -> str:
    drop_prefixes = (
        "Go to Previous Versions",
        "View All Versions",
        "Learn more",
        "This media-neutral citation",
    )
    drop_exact = {
        "of",
        "this Section",
        "Universal Citation:",
        "Next",
        "Previous",
    }
    lines = text.split("\n")
    out: list[str] = []
    skip_until_substantive = True
    for line in lines:
        s = line.strip()
        if not s:
            if not skip_until_substantive:
                out.append("")
            continue
        if s in drop_exact:
            continue
        if any(s.startswith(p) for p in drop_prefixes):
            continue
        if s.startswith("20") and "Massachusetts General Laws" in s:
            continue
        if s.startswith("Disclaimer:") or s.startswith("These codes may not"):
            continue
        skip_until_substantive = False
        out.append(s)
    return "\n".join(out).strip()


def download_insurance_cluster() -> dict[str, int]:
    if REUSE_DISCOVERED_URLS and DISCOVERED_LIST.exists() and DISCOVERED_LIST.stat().st_size > 50:
        raw = [ln.strip() for ln in DISCOVERED_LIST.read_text(encoding="utf-8").splitlines() if ln.strip()]
        all_urls = sorted(raw, key=lambda u: label_sort_key(statute_label_from_url(u)))
        print(f"Loaded {len(all_urls)} section URLs from {DISCOVERED_LIST.name} (skipped discovery)")
    else:
        found = discover_section_urls()
        print(f"Discovered {len(found)} section URLs (GL ch. 175--178A cluster)")
        all_urls = found
        DISCOVERED_LIST.write_text("\n".join(all_urls) + "\n", encoding="utf-8")

    todo = all_urls if not MAX_SECTIONS else all_urls[:MAX_SECTIONS]
    if MAX_SECTIONS:
        print(f"Limited downloads to first {len(todo)} sections (MAX_SECTIONS)")

    wrote = skipped = failed = 0
    for i, sec_url in enumerate(todo, 1):
        label = statute_label_from_url(sec_url)
        dest = OUT_DIR / label_to_filename(label)
        if SKIP_EXISTING and dest.exists() and dest.stat().st_size > 80:
            skipped += 1
        else:
            try:
                html = curl_get(sec_url)
                head_t, body_t = extract_primary_text(html)
                body_t = strip_justia_boilerplate(body_t)
                title = head_t.split("::", 1)[0].strip() if head_t else f"Mass. Gen. Laws {label}"
                md = (
                    f"# {title}\n\n"
                    f"**Massachusetts General Laws -- Part I, Title XXII (insurance ch. 175 et seq.)**\n\n"
                    f"**Source (Justia mirror):** {sec_url}\n\n"
                    f"**Verify on official site:** [malegislature.gov](https://malegislature.gov/Laws/GeneralLaws)\n\n"
                    f"**Section (URL label):** {label}\n\n"
                    f"---\n\n"
                    f"{body_t}\n"
                )
                dest.write_text(md, encoding="utf-8")
                wrote += 1
            except Exception as e:
                print(f"FAIL {label}: {e}")
                failed += 1
        if i % 200 == 0:
            print(f"… {i}/{len(todo)} (wrote={wrote} skipped={skipped} failed={failed})")

    print(f"Done. wrote={wrote} skipped={skipped} failed={failed} → {OUT_DIR.resolve()}")
    return {"wrote": wrote, "skipped": skipped, "failed": failed}


download_insurance_cluster()
```
